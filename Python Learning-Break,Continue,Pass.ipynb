```python
for i in range (1,11):
    if i == 5:
        break
    print(i)
```

```python
#CONTINUE STATEMENT

for i in range (1,11):
    if i == 5:
        continue
    print(i)
    print("Hello")
```

```python
 # PASS statement (used when you want to continue coding and want to come again to write the logic "No error will be Thrown")

for i in range(1,11):
    pass
```

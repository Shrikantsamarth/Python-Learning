## Python supports 3 categories of data types:

- **Basic Types** - (integer, float, complex, boolean and string)
- **Container Types** - (List, Tuples, Sets and Dictionary)
- **User-defined Types** - Class

```python
# integer 
print(4)
print(1e308)
print(1e309) #this is the highest length python can display
```

```python
# float (decimal)
print(4.5)
print(1.7e308) #this is the highest length python can display
```

```python
#Boolean
print(True)
print(False)
```

```python
#complex
print(4+5j)
```

```python
#string
print('Kolkata')
print("Kolkata")
print("""Kolkata""")
```

- Container Types

```python
#list
print([1,2,3,4,5])
```

```python
#tuple
print((1,2,3,4,5))
```

#set
print({1,2,3,4,5})

```python
#dictionary
print({"Name":"Shrikant","Age": 35,"Gender":"Male"})
```
